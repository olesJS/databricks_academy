### Task 3: Create your own medallion schemas, backed by your external location

```python
CATALOG_NAME = 'dbr_dev_ua5816bd'
STORAGE_ACCOUNT = 'dlsua5816bd'
LOGIN = 'oles0305'

adls_url = f'abfss://{LOGIN}@{STORAGE_ACCOUNT}.dfs.core.windows.net'
```

```python
%sql

DESCRIBE EXTERNAL LOCATION oles0305_external_location
-- External location URL: abfss://oles0305@dlsua5816bd.dfs.core.windows.net/
```

```python
layers = ['bronze', 'silver', 'gold']

for layer in layers:
    spark.sql(f"""
              CREATE SCHEMA IF NOT EXISTS {CATALOG_NAME}.{LOGIN}_{layer} 
              MANAGED LOCATION '{adls_url}/{layer}'
            """)
```

```python
spark.sql(f"SHOW SCHEMAS IN {CATALOG_NAME} LIKE '{LOGIN}_*'").show()
```
