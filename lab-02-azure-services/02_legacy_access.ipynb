## Task 5: Connecting to the storage account using a Service Principal

```python
SCOPE_NAME = 'oles0305-kv-scope'

client_id = dbutils.secrets.get(scope=SCOPE_NAME, key="sp-databricks-adls-appid")
client_secret = dbutils.secrets.get(scope=SCOPE_NAME, key="sp-databricks-adls-appkey")
tenant_id = dbutils.secrets.get(scope=SCOPE_NAME, key="tenant-id")
```

```python
STORAGE_ACCOUNT = 'dlsua5816bd'
LOGIN = 'oles0305'
adls_url = f'abfss://{LOGIN}@{STORAGE_ACCOUNT}.dfs.core.windows.net'
```

```python
configs = {
    "fs.azure.account.auth.type": "OAuth",
    "fs.azure.account.oauth.provider.type": "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider",
    "fs.azure.account.oauth2.client.id": client_id,
    "fs.azure.account.oauth2.client.secret": client_secret,
    "fs.azure.account.oauth2.client.endpoint": f"https://login.microsoftonline.com/{tenant_id}/oauth2/token",
}

try:
    dbutils.fs.mount(
        source=adls_url, 
        mount_point=f"/mnt/{LOGIN}_raw", 
        extra_configs=configs)
except Exception as e:
    print(e)
```
