# Data Ingestion Preparation (File Splitting)

```python
CATALOG_NAME = "dbr_dev_ua5816bd"
STORAGE_ACCOUNT = "dlsua5816bd"
LOGIN = "oles0305"

adls_url = f'abfss://{LOGIN}@{STORAGE_ACCOUNT}.dfs.core.windows.net'

source_path = f"{adls_url}/raw_data/results.csv"
target_path = f"{adls_url}/raw_data/raw_results_data"
```

### Loading Baseline CSV Dataset

```python
df = (
    spark.read.format("csv")
        .option("header", True)
        .option("inferSchema", True)
        .option("nullValue", "\\N")
        .load(source_path)
)
```

### Sharding and Exporting to JSON

```python
# Split the DataFrame into 1000 smaller DataFrames
df_split = df.repartition(1000)

(
    df_split.write.format("json")
        .mode("overwrite")
        .save(target_path)
)
```
